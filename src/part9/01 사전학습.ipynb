```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models
```

```python
# 모델 로딩
model_conv = models.resnet18(weights='IMAGENET1K_V1') 
```

```python
print(model_conv) # 모델 구조 출력, 사이즈 작다
```

```python
# 2.모든 파라미터를 고정(False)
for param in model_conv.parameters():
    # print(param)
    param.requires_grad=False
```

```python
#3.마지막 분류 레이어(fc)를 새로운 레이어(Linear) 교체
num_ftrs = model_conv.fc.in_features
# print(num_ftrs)
# model_conv.fc = nn.Linear(in_features=num_ftrs,out_features=2)
model_conv.fc = nn.Linear(num_ftrs,2)
```

```python
#4.모델을 cuda로 이동
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model_conv = model_conv.to(device)
```

```python
# 5.손실함수 정의
criterion = nn.CrossEntropyLoss()
# 6.옵티마이저 정의
optim_conv = optim.SGD(model_conv.fc.parameters(),lr=0.001, momentum=0.9) # fc만 업데이트해~~

# 비교 : vit -> adamw사용
```
